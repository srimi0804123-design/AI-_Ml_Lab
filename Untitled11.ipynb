```python
# Standard Scaler + Normalization using Pandas

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# -----------------------------
# Create 5000 Random Dataset
# -----------------------------
data = {
    "Marks": np.random.randint(1, 100, 5000)
}

df = pd.DataFrame(data)

print(df.head())

# -----------------------------
# Standard Scaling
# -----------------------------
scaler = StandardScaler()

df["Standard_Scaled"] = scaler.fit_transform(df[["Marks"]])

# -----------------------------
# Normalization
# -----------------------------
normalizer = MinMaxScaler()

df["Normalized"] = normalizer.fit_transform(df[["Marks"]])

print(df.head())

# -----------------------------
# Simple Graph
# -----------------------------
plt.plot(df["Marks"][:100], label="Original")
plt.plot(df["Standard_Scaled"][:100], label="Scaled")
plt.plot(df["Normalized"][:100], label="Normalized")

plt.title("Scaling and Normalization")
plt.xlabel("Index")
plt.ylabel("Values")

plt.legend()
plt.show()
```
